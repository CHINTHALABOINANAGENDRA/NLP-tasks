```python
import nltk
import os
from nltk.tokenize import word_tokenize
from transformers import AutoTokenizer

# Set a specific NLTK data path and ensure it exists


nltk_data_path = "/root/nltk_data"
os.makedirs(nltk_data_path, exist_ok=True)
nltk.data.path.append(nltk_data_path)

# Download the 'punkt' and 'punkt_tab' tokenizer data to the specified path
nltk.download('punkt', download_dir=nltk_data_path)
nltk.download('punkt_tab', download_dir=nltk_data_path)

# --- NLTK Tokenization ---
text_nltk = "Tokenization without transformers is straightforward with tools like NLTK."
tokens_nltk = word_tokenize(text_nltk)
print("Tokens:", tokens_nltk)


# --- Transformers Tokenization ---
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
text_transformers = "Tokenization without transformers is straightforward with tools like nltk."
tokens_transformers = tokenizer(text_transformers, return_tensors="pt")

print("Transformers Tokens:", tokens_transformers)

tokens_transformers_list = tokenizer.convert_ids_to_tokens(tokens_transformers['input_ids'][0].numpy().tolist())
print("Transformers Tokens (List):", tokens_transformers_list)

decoded_text = tokenizer.decode(tokens_transformers['input_ids'][0], skip_special_tokens=True)
print("Decoded Text:", decoded_text)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist


nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

def tokenize_document(document):
    tokens = word_tokenize(document)
    return [word.lower() for word in tokens if word.isalpha()]

def remove_stopwords(tokens):
    stop_words = set(stopwords.words('english'))
    return [word for word in tokens if word not in stop_words]

def find_morphology(tokens, top_n=None):
    fdist = FreqDist(tokens)
    return fdist.most_common(top_n) if top_n else fdist.most_common()


user_text = input("Enter your text: ")

tokens = tokenize_document(user_text)
tokens_without_stopwords = remove_stopwords(tokens)
morphology = find_morphology(tokens_without_stopwords, top_n=10)

print("\nTop 10 most common words (excluding stopwords):")
for word, frequency in morphology:
    print(f"{word}: {frequency}")
```
